```python
sns.histplot(df['speed_mph'].dropna(), bins=30)
plt.title('Distribution of Roller Coaster Speeds')
plt.xlabel('Speed (mph)')
plt.ylabel('Number of Coasters')
plt.show()
```

```python
sns.histplot(df['height_ft'].dropna(), bins=30)
plt.title('Distribution of Roller Coaster Heights')
plt.xlabel('Height (feet)')
plt.ylabel('Number of Coasters')
plt.show()
```

```python
sns.scatterplot(data=df, x='height_ft', y='speed_mph')
plt.title('Roller Coaster Height vs Speed')
plt.xlabel('Height (feet)')
plt.ylabel('Speed (mph)')
plt.show()
```

```python
df[['height_ft', 'speed_mph']].corr()
```

```python
df[['coaster_name', 'speed_mph']].dropna().sort_values(
    by='speed_mph', ascending=False
).head(10)
```

```python
top_speed = df[['coaster_name', 'speed_mph']].dropna().sort_values(
    by='speed_mph', ascending=False
).head(10)

sns.barplot(data=top_speed, x='speed_mph', y='coaster_name')

plt.title('Top 10 Fastest Roller Coasters')
plt.xlabel('Speed (mph)')
plt.ylabel('Roller Coaster')
plt.show()
```

```python
df['Manufacturer'].value_counts().head(10)
```

```python
top_manufacturers = df['Manufacturer'].value_counts().head(10)

sns.barplot(x=top_manufacturers.values, y=top_manufacturers.index)

plt.title('Top 10 Roller Coaster Manufacturers')
plt.xlabel('Number of Coasters')
plt.ylabel('Manufacturer')
plt.show()
```

```python
year_counts = df['year_introduced'].value_counts().sort_index()

plt.figure(figsize=(12, 5))
plt.plot(year_counts.index, year_counts.values)

plt.title('Number of Roller Coasters Introduced by Year')
plt.xlabel('Year')
plt.ylabel('Number of Coasters')
plt.show()
```

```python
type_counts = df['Type_Main'].value_counts().head(10)

sns.barplot(x=type_counts.values, y=type_counts.index)

plt.title('Top 10 Roller Coaster Types')
plt.xlabel('Number of Coasters')
plt.ylabel('Coaster Type')
plt.show()
```

```python
inversion_counts = df['Inversions_clean'].value_counts().sort_index()

sns.barplot(x=inversion_counts.index, y=inversion_counts.values)

plt.title('Distribution of Roller Coasters by Number of Inversions')
plt.xlabel('Number of Inversions')
plt.ylabel('Number of Coasters')
plt.show()
```

```python
status_counts = df['Status'].value_counts().head(10)

sns.barplot(x=status_counts.values, y=status_counts.index)

plt.title('Roller Coaster Status Distribution')
plt.xlabel('Number of Coasters')
plt.ylabel('Status')
plt.show()
```

```python
location_counts = df['Location'].value_counts().head(10)

sns.barplot(x=location_counts.values, y=location_counts.index)

plt.title('Top 10 Locations with Most Roller Coasters')
plt.xlabel('Number of Coasters')
plt.ylabel('Location')
plt.show()
```

```python
sns.scatterplot(data=df, x='Inversions_clean', y='speed_mph')

plt.title('Number of Inversions vs Roller Coaster Speed')
plt.xlabel('Number of Inversions')
plt.ylabel('Speed (mph)')
plt.show()
```

```python
sns.histplot(df['Gforce_clean'].dropna(), bins=20)

plt.title('Distribution of Roller Coaster G-Force')
plt.xlabel('G-Force')
plt.ylabel('Number of Coasters')
plt.show()
```

```python
corr = df[['speed_mph', 'height_ft', 'Inversions_clean', 'Gforce_clean']].corr()

sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')

plt.title('Correlation Between Roller Coaster Features')
plt.show()
```

## Key Insights

- The dataset contains 1,087 roller coasters with 56 features.
- Formula Rossa is the fastest roller coaster with a speed of 149.1 mph.
- Vekoma is the most represented manufacturer with 135 roller coasters.
- The average roller coaster speed is approximately 48.6 mph.
- Height and speed have a strong positive correlation of 0.83.
- The dataset contains roller coasters introduced from 1884 to 2022.
- The analysis explored speed, height, manufacturers, coaster types, inversions, status, locations, and G-force.

## Conclusion

The Exploratory Data Analysis helped identify important patterns and relationships in the roller coaster dataset. Visualizations and statistical analysis showed that taller roller coasters generally tend to have higher speeds.
